```typescript
/** @jsxImportSource @nic/openscad-jsx */

import { ComponentChildren, renderToString, type JSX } from "@nic/openscad-jsx";
import { renderToSTL, renderToImage, configureJupyter } from "@nic/openscad-jsx/deno";

configureJupyter({
  viewportHeight: 300,
  backgroundColor: 0xffffff,
  defaultColor: 0xaaaa00,
  gridSize: 300,
});

const $fn = 100;
```

```typescript
const MIN_DIAMETER = 35.5;
const MAX_DIAMETER = 36.5;
const INNER_DIAMETER = 30;
const INNER_HEIGHT = 20;
const BASE_DIAMETER = 44;
const BASE_HEIGHT = 3;
```

```typescript
/** @jsxImportSource @nic/openscad-jsx */

const lid = <>
  <cylinder r={BASE_DIAMETER / 2} h={BASE_HEIGHT} $fn={$fn} />
  <difference>
    <cylinder
      topRadius={MIN_DIAMETER / 2}
      bottomRadius={MAX_DIAMETER / 2}
      h={BASE_HEIGHT + INNER_HEIGHT}
      $fn={$fn}
    />
    <translate z={BASE_HEIGHT + INNER_DIAMETER / 2}>
      <cylinder
        r={INNER_DIAMETER / 2}
        h={INNER_HEIGHT}
      />
      <sphere r={INNER_DIAMETER / 2} />
    </translate>
  </difference>
</>;

const stl = await renderToSTL(lid);

stl;
```

```typescript
await renderToImage(lid);
```

```typescript
Deno.writeFileSync("round_lid.stl", stl);
```
