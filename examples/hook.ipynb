```typescriptreact
/** @jsxImportSource @nic/openscad-jsx */

import { ComponentChildren, renderToString, type JSX } from "@nic/openscad-jsx";
import { renderToSTL, renderToImage, configureJupyter } from "@nic/openscad-jsx/deno";

configureJupyter({
  viewportHeight: 300,
  backgroundColor: 0xffffff,
  defaultColor: 0xaaaa00,
  gridSize: 200,
});

function render2DToStl(element: JSX.Element) {
  return renderToSTL(<extrusionLinear h={0.01}>{element}</extrusionLinear>);
}
```

```typescriptreact
/** @jsxImportSource @nic/openscad-jsx */

//  +----b-----+
//  a          |
//  |          |
//             c      e
//             |      |
//             +---d--+

const a = <rectangle sx={2} sy={40} />;
const b = <rectangle sx={39.5 + 2} sy={2} />;
const c = <rectangle sx={2} sy={30} />;
const e = <rectangle sx={2} sy={5} />;

function HalfCircle({ r, extra = 0 }: { r: number, extra?: number }) {
  return (
    <intersection>
      <translate x={r} y={r}>
        <circle r={r} $fn={1000} />
      </translate>
      <rectangle sx={2 * r} sy={r + extra} />
    </intersection>
  )
}

const d = <difference>
  <HalfCircle r={7} />
  <translate x={7} y={7}>
    <circle r={5} $fn={1000} />
  </translate>
</difference>;

const proj = <>
  <translate x={-2} y={0}>{a}</translate>
  <translate x={-2} y={38}>{b}</translate>
  <translate x={39.5} y={10}>{c}</translate>
  <translate x={39.5} y={3}>{d}</translate>
  <translate x={39.5 + 12} y={10}>{e}</translate>
</>;

await renderToImage(proj);
```

```typescriptreact
/** @jsxImportSource @nic/openscad-jsx */

const hook = <extrusionLinear h={10}>{proj}</extrusionLinear>;

const stl = await renderToSTL(hook);

stl;
```
